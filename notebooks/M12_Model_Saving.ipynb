```python
from pathlib import Path
import joblib

ROOT = Path(r"..")
MODEL_DIR = ROOT / "backend" / "trained_models"
MODEL_DIR.mkdir(parents=True, exist_ok=True)

# Load the Random Forest model if needed
rf_model = joblib.load(
    MODEL_DIR / "random_forest.pkl"
)

# Save as the current production candidate
production_model_path = MODEL_DIR / "authentiscore_ensemble_v3.pkl"

joblib.dump(
    rf_model,
    production_model_path
)

print("✅ Production candidate saved:")
print(production_model_path)
```

```python
import joblib

loaded_model = joblib.load(
    production_model_path
)

print("Model type:", type(loaded_model).__name__)
print("Number of trees:", loaded_model.n_estimators)
```
