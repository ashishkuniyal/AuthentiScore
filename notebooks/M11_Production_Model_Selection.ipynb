```python
from pathlib import Path
import pandas as pd

ROOT = Path(r"..")

results = pd.read_csv(
    ROOT / "data" / "processed" / "model_comparison.csv"
)

print(results.round(4))
```

```python
best_auc_model = results.loc[
    results["ROC-AUC"].idxmax(),
    "Model"
]

best_f1_model = results.loc[
    results["F1"].idxmax(),
    "Model"
]

print("Highest ROC-AUC:", best_auc_model)
print("Highest F1:", best_f1_model)
```
