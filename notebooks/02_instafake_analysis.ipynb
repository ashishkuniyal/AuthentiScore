```python
import pandas as pd
import numpy as np
import json
import os
```

```python
FAKE_PATH = "../data/raw/instafake/fake-v1.0/fakeAccountData.json"
REAL_PATH = "../data/raw/instafake/fake-v1.0/realAccountData.json"

print("Fake file:")
print(os.path.abspath(FAKE_PATH))

print("\nReal file:")
print(os.path.abspath(REAL_PATH))
```

```python
print("Fake file exists:", os.path.exists(FAKE_PATH))
print("Real file exists:", os.path.exists(REAL_PATH))
```

```python
with open(FAKE_PATH, "r", encoding="utf-8") as f:
    fake_data = json.load(f)

with open(REAL_PATH, "r", encoding="utf-8") as f:
    real_data = json.load(f)

print("Fake records:", len(fake_data))
print("Real records:", len(real_data))
```

```python
print("Type of fake_data:", type(fake_data))
print("Type of real_data:", type(real_data))
print("\nFirst fake record:")
print(fake_data[0])
print("\nFirst real record:")
print(real_data[0])
```

```python
fake_df = pd.DataFrame(fake_data)
real_df = pd.DataFrame(real_data)

print("Fake DataFrame shape:", fake_df.shape)
print("Real DataFrame shape:", real_df.shape)
```

```python
fake_df["fake"] = 1
real_df["fake"] = 0
print("Fake dataset:")
print(fake_df.shape)

print("\nReal dataset:")
print(real_df.shape)
```

```python
instafake_df = pd.concat(
    [fake_df, real_df],
    ignore_index=True
)

print("Combined InstaFake shape:")
print(instafake_df.shape)
```

```python
print("InstaFake Columns")
print("=================")

for i, column in enumerate(instafake_df.columns, start=1):
    print(f"{i}. {column}")
```

```python
print(instafake_df.dtypes)
```

```python
instafake_df.info()
```

```python
missing_values = instafake_df.isnull().sum()

print("Missing Values")
print("==============")

print(missing_values)
```

```python
missing_report = pd.DataFrame({
    "Column": instafake_df.columns,
    "Missing Values": instafake_df.isnull().sum().values,
    "Missing %": (
        instafake_df.isnull().mean() * 100
    ).round(2).values
})

missing_report
```

```python
duplicate_count = instafake_df.duplicated().sum()

print("Duplicate rows:", duplicate_count)
```

```python
print("Target Distribution")
print("===================")

print(instafake_df["fake"].value_counts())
```

```python
print("Target Percentage")
print("=================")

print(
    instafake_df["fake"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)
```

```python
dataset_report = pd.DataFrame({
    "Column": instafake_df.columns,
    "Data Type": [
        str(instafake_df[column].dtype)
        for column in instafake_df.columns
    ],
    "Missing Values": [
        instafake_df[column].isnull().sum()
        for column in instafake_df.columns
    ],
    "Missing %": [
        round(
            instafake_df[column].isnull().mean() * 100,
            2
        )
        for column in instafake_df.columns
    ],
    "Unique Values": [
        instafake_df[column].nunique()
        for column in instafake_df.columns
    ]
})

dataset_report
```

```python
REPORT_PATH = "../data/processed/instafake_inspection_report.csv"

dataset_report.to_csv(
    REPORT_PATH,
    index=False
)

print("✅ Report saved:")
print(REPORT_PATH)
```
