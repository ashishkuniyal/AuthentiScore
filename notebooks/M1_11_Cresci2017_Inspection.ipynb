```python
import pandas as pd
import numpy as np
from pathlib import Path
```

```python
# Project root
PROJECT_ROOT = Path.cwd().parent

# Cresci-2017 dataset
DATA_DIR = PROJECT_ROOT / "data" / "external" / "cresci2017"

genuine_path = DATA_DIR / "genuine_users.csv"
spam_path = DATA_DIR / "spam_user.csv"

print("Genuine file:", genuine_path)
print("Spam file:", spam_path)

print("\nGenuine exists:", genuine_path.exists())
print("Spam exists:", spam_path.exists())
```

```python
genuine_df = pd.read_csv(genuine_path)
spam_df = pd.read_csv(spam_path)

print("Genuine dataset shape:", genuine_df.shape)
print("Spam dataset shape:", spam_df.shape)
```

```python
print("Genuine columns:")
print(genuine_df.columns.tolist())

print("\nSpam columns:")
print(spam_df.columns.tolist())
```
