```python
import re
import pandas as pd

RISK_PATTERNS = {
    "scam": [
        r"\bscam\b",
        r"\bscammer\b",
        r"\bscammed\b",
        r"\bfraud\b",
        r"\bfraudulent\b"
    ],

    "money": [
        r"\bfree money\b",
        r"\bearn money\b",
        r"\bmake money\b",
        r"\bcash\b",
        r"\bprofit\b",
        r"\binvest\b",
        r"\binvestment\b"
    ],

    "urgency": [
        r"\burgent\b",
        r"\bact now\b",
        r"\blimited time\b",
        r"\blast chance\b"
    ],

    "contact": [
        r"\bcontact me\b",
        r"\bdm me\b",
        r"\bdirect message\b",
        r"\bwhatsapp\b",
        r"\btelegram\b"
    ],

    "promotional": [
        r"\bgiveaway\b",
        r"\bfree\b",
        r"\bwin\b",
        r"\bbonus\b",
        r"\boffer\b",
        r"\bdiscount\b"
    ]
}

print("Risk categories:", list(RISK_PATTERNS.keys()))
```

```python
def detect_risk_patterns(text):
    if pd.isna(text):
        text = ""

    text = str(text).lower()

    matches = []

    for category, patterns in RISK_PATTERNS.items():
        for pattern in patterns:
            found = re.findall(pattern, text)

            for match in found:
                matches.append({
                    "category": category,
                    "matched_text": match
                })

    return matches
```

```python
def calculate_nlp_rule_score(text):
    matches = detect_risk_patterns(text)

    category_weights = {
        "scam": 30,
        "money": 20,
        "urgency": 20,
        "contact": 15,
        "promotional": 15
    }

    score = 0

    for match in matches:
        score += category_weights.get(
            match["category"],
            0
        )

    score = min(score, 100)

    return score, matches
```

```python
test_bios = [
    "Win free money! DM me on Telegram",
    "Computer science student",
    "Limited time investment offer, contact me now",
    "Travel | Photography | Mountains"
]

for bio in test_bios:
    score, matches = calculate_nlp_rule_score(bio)

    print("\nBio:", bio)
    print("Risk score:", score)
    print("Matches:", matches)
```
