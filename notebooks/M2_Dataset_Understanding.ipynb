```python
import pandas as pd
from pathlib import Path

ROOT = Path.cwd().parent

# =========================
# Dataset A - Instagram
# =========================
TRAIN_PATH = ROOT / "data" / "raw" / "instagram_fake_spammer" / "train.csv"
TEST_PATH = ROOT / "data" / "raw" / "instagram_fake_spammer" / "test.csv"

train_df = pd.read_csv(TRAIN_PATH)
test_df = pd.read_csv(TEST_PATH)

dataset_a = pd.concat([train_df, test_df], ignore_index=True)

# =========================
# Dataset B - InstaFake
# =========================
fake_path = ROOT / "data" / "raw" / "instafake" / "fake-v1.0" / "fakeAccountData.json"
real_path = ROOT / "data" / "raw" / "instafake" / "fake-v1.0" / "realAccountData.json"

fake_df = pd.read_json(fake_path)
real_df = pd.read_json(real_path)

# =========================
# Dataset C - Cresci-2017
# =========================
genuine_path = ROOT / "data" / "external" / "cresci2017" / "genuine_users.csv"
spam_path = ROOT / "data" / "external" / "cresci2017" / "spam_user.csv"

genuine_df = pd.read_csv(genuine_path)
spam_df = pd.read_csv(spam_path)

# =========================
# Display shapes
# =========================
print("Dataset A - train:", train_df.shape)
print("Dataset A - test :", test_df.shape)
print("Dataset A - total:", dataset_a.shape)

print("\nInstaFake - fake :", fake_df.shape)
print("InstaFake - real :", real_df.shape)

print("\nCresci - genuine:", genuine_df.shape)
print("Cresci - spam   :", spam_df.shape)
```

```python
print("=" * 60)
print("DATASET A - INSTAGRAM")
print("=" * 60)

print("Columns:")
print(dataset_a.columns.tolist())

print("\nTarget distribution:")
print(dataset_a["fake"].value_counts(dropna=False))


print("\n" + "=" * 60)
print("INSTaFAKE")
print("=" * 60)

print("Columns:")
print(fake_df.columns.tolist())

print("\nFake accounts - isFake:")
print(fake_df["isFake"].value_counts(dropna=False))

print("\nReal accounts - isFake:")
print(real_df["isFake"].value_counts(dropna=False))


print("\n" + "=" * 60)
print("CRESCI")
print("=" * 60)

print("Genuine shape:", genuine_df.shape)
print("Spam shape:", spam_df.shape)

print("\nGenuine columns:")
print(genuine_df.columns.tolist())

print("\nSpam columns:")
print(spam_df.columns.tolist())
```
