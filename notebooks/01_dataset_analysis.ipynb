```python
# ============================================
# M1.5 - DATASET INSPECTION
# AuthentiScore
# ============================================

import pandas as pd
import numpy as np
import os
import glob
```

```python
# Path to the raw dataset

DATA_DIR = "../data/raw/instagram_fake_spammer"

print("Dataset directory:")
print(os.path.abspath(DATA_DIR))
```

```python
# Check if dataset folder exists

if os.path.exists(DATA_DIR):
    print("✅ Dataset folder found!")
else:
    print("❌ Dataset folder NOT found.")
    print("Check the DATA_DIR path.")
```

```python
# Display files inside the dataset folder

files = os.listdir(DATA_DIR)

print("Files found:")
print("----------------")

for file in files:
    print(file)
```

```python
# Find all CSV files

csv_files = glob.glob(os.path.join(DATA_DIR, "*.csv"))

print(f"Number of CSV files found: {len(csv_files)}")
print()

for file in csv_files:
    print(file)
```

```python
# Check size of each CSV file

for file in csv_files:
    size_mb = os.path.getsize(file) / (1024 * 1024)

    print(f"{os.path.basename(file)} : {size_mb:.2f} MB")
```

```python
TRAIN_PATH = os.path.join(DATA_DIR, "train.csv")

train_df = pd.read_csv(TRAIN_PATH)

print("✅ Training dataset loaded successfully!")
```

```python
TEST_PATH = os.path.join(DATA_DIR, "test.csv")

if os.path.exists(TEST_PATH):

    test_df = pd.read_csv(TEST_PATH)

    print("✅ Test dataset loaded successfully!")

else:

    test_df = None

    print("⚠️ test.csv not found.")
```

```python
print("Training Dataset Shape")
print("======================")
print(f"Rows    : {train_df.shape[0]}")
print(f"Columns : {train_df.shape[1]}")
```

```python
if test_df is not None:

    print("Test Dataset Shape")
    print("==================")
    print(f"Rows    : {test_df.shape[0]}")
    print(f"Columns : {test_df.shape[1]}")

else:

    print("No separate test dataset found.")
```

```python
missing_percentage = (
    train_df.isnull().sum()
    / len(train_df)
    * 100
)

missing_report = pd.DataFrame({
    "Missing Count": train_df.isnull().sum(),
    "Missing Percentage": missing_percentage
})

missing_report
```

```python
duplicate_count = train_df.duplicated().sum()

print("Duplicate Rows:", duplicate_count)
```

```python
duplicates = train_df[train_df.duplicated(keep=False)]

print("Number of rows involved in duplicates:", len(duplicates))

duplicates.head(20)
```

```python
unique_report = pd.DataFrame({
    "Column": train_df.columns,
    "Unique Values": [
        train_df[column].nunique()
        for column in train_df.columns
    ]
})

unique_report
```

```python
if "Fake" in train_df.columns:

    print("✅ 'Fake' target column found!")

else:

    print("⚠️ 'Fake' column not found.")

    print("\nAvailable columns:")
    print(train_df.columns.tolist())
```

```python
if "Fake" in train_df.columns:

    target_counts = train_df["Fake"].value_counts()

    print("Target Distribution")
    print("===================")

    print(target_counts)
```

```python
if "Fake" in train_df.columns:

    target_percentage = (
        train_df["Fake"]
        .value_counts(normalize=True)
        * 100
    )

    print("Target Percentage")
    print("=================")

    print(target_percentage)
```

```python
if "Fake" in train_df.columns:

    target_summary = pd.DataFrame({
        "Count": train_df["Fake"].value_counts(),
        "Percentage": (
            train_df["Fake"]
            .value_counts(normalize=True)
            * 100
        ).round(2)
    })

    target_summary
```

```python
for column in train_df.columns:

    print("\n" + "=" * 60)
    print(f"COLUMN: {column}")
    print("=" * 60)

    print(train_df[column].head(10).to_string(index=False))
```

```python
numeric_columns = train_df.select_dtypes(
    include=np.number
).columns.tolist()

print("Numerical Columns")
print("=================")

for column in numeric_columns:
    print("-", column)
```

```python
categorical_columns = train_df.select_dtypes(
    exclude=np.number
).columns.tolist()

print("Categorical / Text Columns")
print("=========================")

for column in categorical_columns:
    print("-", column)
```

```python
if len(numeric_columns) > 0:

    numeric_summary = pd.DataFrame({
        "Minimum": train_df[numeric_columns].min(),
        "Maximum": train_df[numeric_columns].max(),
        "Mean": train_df[numeric_columns].mean(),
        "Median": train_df[numeric_columns].median()
    })

    numeric_summary
```

```python
if len(numeric_columns) > 1:

    correlation = train_df[numeric_columns].corr()

    correlation
```

```python
dataset_report = pd.DataFrame({
    "Column": train_df.columns,
    "Data Type": [
        str(train_df[column].dtype)
        for column in train_df.columns
    ],
    "Missing Values": [
        train_df[column].isnull().sum()
        for column in train_df.columns
    ],
    "Missing %": [
        round(
            train_df[column].isnull().mean() * 100,
            2
        )
        for column in train_df.columns
    ],
    "Unique Values": [
        train_df[column].nunique()
        for column in train_df.columns
    ]
})

dataset_report
```

```python
print("=" * 60)
print("        AUTHENTISCORE - DATASET INSPECTION")
print("=" * 60)

print("\nDataset:")
print("Instagram Fake Spammer / Genuine Accounts")

print("\nRows:", train_df.shape[0])
print("Columns:", train_df.shape[1])

print("\nDuplicate Rows:", train_df.duplicated().sum())

print("\nTotal Missing Values:",
      train_df.isnull().sum().sum())

print("\nNumerical Columns:",
      len(numeric_columns))

print("Text/Categorical Columns:",
      len(categorical_columns))

print("\nColumn Names:")
for column in train_df.columns:
    print("  -", column)

if "Fake" in train_df.columns:

    print("\nTarget Distribution:")
    print(train_df["Fake"].value_counts())

print("\n" + "=" * 60)
print("        INSPECTION COMPLETE")
print("=" * 60)
```

```python
REPORT_PATH = "../data/processed/dataset_inspection_report.csv"

dataset_report.to_csv(
    REPORT_PATH,
    index=False
)

print("✅ Inspection report saved to:")
print(REPORT_PATH)
```
